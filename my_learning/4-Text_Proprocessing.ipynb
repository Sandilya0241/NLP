```python
corpus = """Hello Welcome, to Sandilya Sripathi's NLP learning.
Please do watch the entire course! to become expert in NLP."""

from nltk.tokenize import sent_tokenize
# Below functionality will convert the corpus(paragraphs) into documents(sentences)
sentences = sent_tokenize(corpus)
type(sentences)
```

```python
for sentence in sentences:
    print(sentence)
```

#### II. Tokenization (we can convert corpus/ paragrah the below ways:)
##### 1. corpus -> documents (sentences)
##### 2. corpus (paragragh) -> words

```python
corpus = """Hello Welcome, to Sandilya Sripathi's NLP learning.
Please do watch the entire course! to become expert in NLP."""
from nltk.tokenize import word_tokenize

words = word_tokenize(corpus)
words
```

```python
## Sentence -> Word Tokenization
for sentence in sentences:
    print(word_tokenize(sentence))
```

```python
corpus = """Hello Welcome, to Sandilya Sripathi's NLP learning.
Please do watch the entire course! to become expert in NLP."""
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
corpus = """Hello Welcome, to Sandilya Sripathi's NLP learning.
Please do watch the entire course! to become expert in NLP."""
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)

## difference would be that it will not split the word "NLP's" into "NLP" and "'s" but will keep it as one word. Also it will not consider . as a separate token except when it's at the end of a sentence.
```
